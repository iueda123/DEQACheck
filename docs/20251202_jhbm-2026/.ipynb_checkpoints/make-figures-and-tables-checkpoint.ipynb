## チャンク 1: データ前処理

```r
# ---- 1.1 入力ファイル読み込み ----
raw <- read_tsv("figs/record_ver20260313.csv",
                col_types = cols(.default = "c"))
pub  <- read_tsv("figs/pubdate_by_authorYear.tsv",
                 col_types = cols(.default = "c"))

cat("raw行数:", nrow(raw), "\n")
cat("pub行数:", nrow(pub), "\n")

# ---- 1.2 ゴミ行・重複の除去 ----
# ゴミ行除去
raw2 <- raw %>%
  filter(!authorYear %in% c("settings", "Someone20XX"))

# phase 優先度: Train=1, Overall=2, Uninvestigated=3, その他=4
phase_priority <- function(p) {
  case_when(
    p == "Train"          ~ 1L,
    p == "Overall"        ~ 2L,
    p == "Uninvestigated" ~ 3L,
    TRUE                  ~ 4L
  )
}

df <- raw2 %>%
  mutate(phase_rank = phase_priority(phase)) %>%
  group_by(authorYear) %>%
  slice_min(phase_rank, n = 1, with_ties = FALSE) %>%
  ungroup() %>%
  select(-phase_rank)

cat("df（重複除去後）行数:", nrow(df), "\n")

# ---- 1.3 モダリティ正規化 ----
# セミコロン分割 → 各トークンを4カテゴリに変換
normalize_modality <- function(token) {
  token <- str_trim(token)
  case_when(
    token %in% c("T1w MRI", "T2w MRI")         ~ "sMRI",
    token == "fMRI"                              ~ "fMRI",
    token == "dMRI"                              ~ "dMRI",
    TRUE                                         ~ "Other"
  )
}

df_mod <- df %>%
  select(authorYear, modality) %>%
  mutate(modality_tokens = str_split(modality, ";")) %>%
  unnest(modality_tokens) %>%
  mutate(modality_tokens = str_trim(modality_tokens)) %>%
  filter(modality_tokens != "") %>%
  mutate(modality_cat = normalize_modality(modality_tokens)) %>%
  distinct(authorYear, modality_cat)

cat("df_mod 行数:", nrow(df_mod), "（1文献×1カテゴリ）\n")
print(table(df_mod$modality_cat))

# ---- 1.4 疾患名正規化 ----
# 括弧内の略語のみを取り出す（例: Schizophrenia (SCZ) → SCZ）
# 括弧がない場合はそのまま
normalize_disease <- function(token) {
  token <- str_trim(token)
  # 括弧付き略語を抽出
  abbr <- str_extract(token, "(?<=\\()([^)]+)(?=\\))")
  # 括弧内が空または存在しない場合はトークンそのまま
  if_else(!is.na(abbr) & abbr != "", abbr, token)
}

df_dis <- df %>%
  select(authorYear, disease) %>%
  mutate(disease_tokens = str_split(disease, ";")) %>%
  unnest(disease_tokens) %>%
  mutate(disease_tokens = str_trim(disease_tokens)) %>%
  filter(disease_tokens != "") %>%
  mutate(disease_norm = normalize_disease(disease_tokens)) %>%
  distinct(authorYear, disease_norm)

cat("df_dis 行数:", nrow(df_dis), "（1文献×1疾患）\n")
print(sort(table(df_dis$disease_norm), decreasing = TRUE)[1:15])

# ---- 1.5 出版年の取得 ----
# pub の pub_year, pub_month を数値に変換
pub2 <- pub %>%
  mutate(
    pub_year  = as.integer(pub_year),
    pub_month = as.integer(pub_month),
    # TSV内 pub_month 欠損 4件を 12 で補完
    pub_month = if_else(is.na(pub_month), 12L, pub_month)
  )

# df に left_join
df <- df %>%
  left_join(pub2 %>% select(authorYear, pub_year, pub_month), by = "authorYear")

# フォールバック: TSVにない22件は authorYear 末尾4桁から pub_year 補完
extract_year_from_authorYear <- function(ay) {
  # 末尾の4桁数字を取り出す（A/Bサフィックスは除去）
  yr_str <- str_extract(ay, "\\d{4}(?=[AB]?$)")
  as.integer(yr_str)
}

df <- df %>%
  mutate(
    pub_year  = if_else(is.na(pub_year),  extract_year_from_authorYear(authorYear), pub_year),
    pub_month = if_else(is.na(pub_month), 12L, pub_month)
  )

cat("pub_year 欠損数:", sum(is.na(df$pub_year)), "\n")
cat("pub_month 欠損数:", sum(is.na(df$pub_month)), "\n")

# N を数値に変換
df <- df %>%
  mutate(
    N          = as.numeric(N),
    age_mean   = as.numeric(age_mean),
    female_pct = as.numeric(female_pct)
  )

cat("\ndf 列名:", paste(names(df), collapse=", "), "\n")
cat("df 行数（最終）:", nrow(df), "\n")
```

## チャンク 2: Fig.1 研究数の四半期別推移

```r
# ---- 四半期ラベル生成 ----
df_fig1 <- df %>%
  mutate(
    quarter = ceiling(pub_month / 3),
    pub_quarter = paste0(pub_year, " Q", quarter)
  )

# モダリティカテゴリを付与（代表カテゴリ: 優先度 sMRI > fMRI > dMRI > Other）
mod_priority <- c("sMRI", "fMRI", "dMRI", "Other")

df_mod_rep <- df_mod %>%
  mutate(mod_rank = match(modality_cat, mod_priority)) %>%
  group_by(authorYear) %>%
  slice_min(mod_rank, n = 1, with_ties = FALSE) %>%
  ungroup() %>%
  select(authorYear, modality_cat)

df_fig1 <- df_fig1 %>%
  left_join(df_mod_rep, by = "authorYear")

# ---- 全四半期の完全シーケンスを生成（等間隔X軸のため） ----
yr_min <- min(df_fig1$pub_year)
yr_max <- max(df_fig1$pub_year)

all_quarters <- expand.grid(
  year    = yr_min:yr_max,
  quarter = 1:4
) %>%
  arrange(year, quarter) %>%
  mutate(pub_quarter = paste0(year, " Q", quarter)) %>%
  pull(pub_quarter)

# 累積積み上げ棒グラフ用カウント（欠損四半期は 0 で補完）
df_fig1_count <- df_fig1 %>%
  count(pub_quarter, modality_cat) %>%
  complete(pub_quarter = all_quarters, modality_cat = mod_priority, fill = list(n = 0L)) %>%
  mutate(
    pub_quarter  = factor(pub_quarter, levels = all_quarters),
    modality_cat = factor(modality_cat, levels = rev(mod_priority))
  ) %>%
  arrange(pub_quarter)

# 累積カウント（各四半期までの累積）
df_fig1_cum <- df_fig1_count %>%
  group_by(modality_cat) %>%
  mutate(cum_n = cumsum(n)) %>%
  ungroup()

# カラーパレット
mod_colors <- c(
  "sMRI"  = "#4472C4",
  "fMRI"  = "#ED7D31",
  "dMRI"  = "#A9D18E",
  "Other" = "#BFBFBF"
)

# x軸ラベル: 各年の Q1 のみ年を表示、他は空白
x_show <- setNames(
  ifelse(grepl(" Q1$", all_quarters), str_extract(all_quarters, "^\\d{4}"), ""),
  all_quarters
)

p1 <- ggplot(df_fig1_cum,
             aes(x = pub_quarter, y = cum_n, fill = modality_cat)) +
  geom_bar(stat = "identity", position = "stack", width = 0.8) +
  scale_fill_manual(
    values = mod_colors,
    breaks = mod_priority,
    labels = mod_priority,
    name   = "Modality"
  ) +
  scale_x_discrete(labels = x_show) +
  labs(
    x        = "Quarter",
    y        = "Cumulative number of studies",
    caption  = "n = 122; pub_month missing → Q4"
  ) +
  theme_classic(base_size = 14) +
  theme(
    axis.text.x  = element_text(angle = 45, hjust = 1, size = 9),
    legend.position = "right",
    plot.title   = element_text(size = 12, face = "bold")
  )

ggsave("figs/fig_trend_by_quarter.png", p1,
       width = 12, height = 4, dpi = 300, bg = "white")
cat("Fig.1 saved\n")
p1
```

## チャンク 9: Table 1 研究サマリー統計表

```r
# モダリティ件数
mod_counts <- df_mod %>%
  count(modality_cat, name = "n")
get_mod <- function(cat) {
  mod_counts %>% filter(modality_cat == cat) %>% pull(n) %>% { if (length(.) == 0) 0L else . }
}

# Origin 件数
origin_counts <- df %>% count(origin, name = "n")
get_origin <- function(ori) {
  origin_counts %>% filter(origin == ori) %>% pull(n) %>% { if (length(.) == 0) 0L else . }
}

# テーブルデータ作成
tbl_data <- tribble(
  ~Category,                     ~Item,              ~Value,
  "Studies",         "Total",            as.character(nrow(df)),
  "Reference cohort","N: median",        as.character(round(median(df$N, na.rm=TRUE))),
  "Reference cohort","N: IQR",           sprintf("%.0f–%.0f",
                                                  quantile(df$N, .25, na.rm=TRUE),
                                                  quantile(df$N, .75, na.rm=TRUE)),
  "Reference cohort","Age (mean): median",sprintf("%.2f yrs", median(df$age_mean, na.rm=TRUE)),
  "Reference cohort","Age (mean): IQR",  sprintf("%.2f–%.2f yrs",
                                                  quantile(df$age_mean, .25, na.rm=TRUE),
                                                  quantile(df$age_mean, .75, na.rm=TRUE)),
  "Reference cohort","Female: median",   sprintf("%.1f%%", median(df$female_pct, na.rm=TRUE)),
  "Reference cohort","Female: IQR",      sprintf("%.1f–%.1f%%",
                                                  quantile(df$female_pct, .25, na.rm=TRUE),
                                                  quantile(df$female_pct, .75, na.rm=TRUE)),
  "Modality",        "sMRI",             as.character(get_mod("sMRI")),
  "Modality",        "fMRI",             as.character(get_mod("fMRI")),
  "Modality",        "dMRI",             as.character(get_mod("dMRI")),
  "Modality",        "Other",            as.character(get_mod("Other")),
  "Model Origin",    "New",              as.character(get_origin("New")),
  "Model Origin",    "Pre-trained",      as.character(get_origin("Pre-trained"))
)

tbl <- tbl_data %>%
  gt(groupname_col = "Category") %>%
  cols_label(Item = "Item", Value = "Value") %>%
  tab_header(
    title    = "Table 1: Summary statistics of NM studies",
    subtitle = "n = 122 studies (2005–2025)"
  ) %>%
  tab_options(
    table.font.size        = 14,
    heading.title.font.size = 16,
    row_group.font.weight  = "bold"
  )

# PNG エクスポート
gtsave(tbl, "figs/table1_summary.png")
cat("Table 1 saved\n")
tbl
```

## チャンク 11: df_mod_dis の生成（df_mod × df_dis）

```r
# ---- Chunk 11: df_mod_dis ----
# df_mod（1行 = 1文献×1モダリティ）と df_dis（1行 = 1文献×1疾患）を
# authorYear で inner_join して 1行 = 1文献×1モダリティ×1疾患 の長形式を作成。
# 1文献が複数モダリティ×複数疾患を持つ場合は全組み合わせが生成される。

df_mod_dis <- df_mod %>%
  inner_join(df_dis, by = "authorYear", relationship = "many-to-many") %>%
  distinct(authorYear, modality_cat, disease_norm)

cat("df_mod_dis 行数:", nrow(df_mod_dis), "\n")
cat("ユニーク文献数:", n_distinct(df_mod_dis$authorYear), "\n")
```

## チャンク 12: Fig.3b 疾患別研究件数（モダリティ色分け積み上げ横棒）

```r
# ---- Chunk 12: Fig.3b ----
# Fig.3 の上位 10 疾患と同じセットを使用してモダリティ色分け積み上げ棒グラフを生成。

# 上位10疾患をdf_mod_disの合計件数で並び替え（y軸の順序をグラフの実合計に一致させる）
top10_order <- df_mod_dis %>%
  count(disease_norm, name = "total") %>%
  slice_max(total, n = 10, with_ties = FALSE) %>%
  arrange(total) %>%
  pull(disease_norm)

# df_mod_dis から上位10疾患を絞り込んでカウント
df_fig3b <- df_mod_dis %>%
  filter(disease_norm %in% top10_order) %>%
  count(disease_norm, modality_cat, name = "n") %>%
  mutate(
    disease_norm = factor(disease_norm, levels = top10_order),
    modality_cat = factor(modality_cat, levels = c("sMRI", "fMRI", "dMRI", "Other"))
  )

cat("Fig.3b データ確認:\n")
print(df_fig3b %>% group_by(disease_norm) %>% summarise(total = sum(n)) %>% arrange(desc(total)))

p3b <- ggplot(df_fig3b, aes(x = n, y = disease_norm, fill = modality_cat)) +
  geom_bar(stat = "identity", position = "stack", width = 0.6) +
  geom_text(
    data = df_fig3b %>%
      group_by(disease_norm) %>%
      summarise(total = sum(n), .groups = "drop"),
    aes(x = total, y = disease_norm, label = total),
    inherit.aes = FALSE,
    hjust = -0.2, size = 4.5, fontface = "bold"
  ) +
  scale_fill_manual(
    values = mod_colors,
    breaks = c("sMRI", "fMRI", "dMRI", "Other"),
    name   = "Modality"
  ) +
  scale_x_continuous(expand = expansion(mult = c(0, 0.15))) +
  labs(
    x       = "Number of studies",
    y       = NULL,
    caption = "Studies with multiple modalities × diseases are counted in each combination."
  ) +
  theme_classic(base_size = 14)

ggsave("figs/fig_disease_modality_stacked.png", p3b,
       width = 9, height = 2.5, dpi = 300, bg = "white")
cat("Fig.3b saved\n")
p3b
```

## チャンク 13: Table 2 Modality × Disease クロス集計表

```r
# ---- Chunk 13: Table 2 ----
# モダリティ（行） × 疾患グループ（列）のクロス集計表。
# BPCNPNP2025_Takamatsu スライドの形式に準拠。

# 疾患グループ定義
df_cross_raw <- df_mod_dis %>%
  mutate(disease_group = case_when(
    disease_norm %in% c("SCZ", "FEP", "CHR-P", "EP") ~ "SCZ系",
    disease_norm == "ASD"                              ~ "ASD",
    disease_norm == "MDD"                              ~ "MDD",
    disease_norm == "BD"                               ~ "BD",
    disease_norm == "ADHD"                             ~ "ADHD",
    disease_norm %in% c("AD", "MCI")                  ~ "AD/MCI",
    disease_norm == "PD"                               ~ "PD",
    disease_norm == "MS"                               ~ "MS",
    TRUE                                               ~ "Other"
  ))

# クロス集計してワイド形式に変換
col_order <- c("SCZ系", "ASD", "MDD", "BD", "ADHD", "AD/MCI", "PD", "MS", "Other")

df_cross <- df_cross_raw %>%
  count(modality_cat, disease_group, name = "n") %>%
  pivot_wider(names_from = disease_group, values_from = n, values_fill = 0L) %>%
  # 列を指定順に並べ、存在しない列は 0 で補完
  mutate(across(any_of(col_order[!col_order %in% names(.)]),
                ~0L, .names = "{.col}")) %>%
  select(Modality = modality_cat, any_of(col_order)) %>%
  mutate(Total = rowSums(across(where(is.integer))))

# モダリティ行を合計件数降順に並べ替え
df_cross <- df_cross %>%
  arrange(desc(Total))

cat("Table 2 クロス集計:\n")
print(df_cross)

tbl2 <- df_cross %>%
  gt() %>%
  tab_header(
    title    = "Table 2: Modality × Disease cross-tabulation",
    subtitle = "n = 122; each study counted in every modality-disease combination used"
  ) %>%
  tab_spanner(
    label   = "Psychiatric",
    columns = any_of(c("SCZ系", "ASD", "MDD", "BD", "ADHD"))
  ) %>%
  tab_spanner(
    label   = "Neurological",
    columns = any_of(c("AD/MCI", "PD", "MS"))
  ) %>%
  tab_spanner(
    label   = "Other/Mixed",
    columns = any_of(c("Other"))
  ) %>%
  tab_style(
    style = cell_text(weight = "bold"),
    locations = cells_column_labels()
  ) %>%
  tab_style(
    style = cell_fill(color = "#F0F0F0"),
    locations = cells_column_spanners()
  ) %>%
  tab_options(
    table.font.size        = 13,
    heading.title.font.size = 15
  )

gtsave(tbl2, "figs/table_modality_disease.png")
cat("Table 2 saved\n")
tbl2
```
